### Celda 1: la campaña del ICT

#### Lo que deben obtener
El terco pierde 112 clientes y le muestra el mejor anuncio al 90.7% de los turistas. El
apostador pierde 25 y llega al 97.7%.
Comparen ese 90.7% con lo que predijeron en A.2b: no es casualidad, es exactamente 90% +
10%/4. Y miren la línea roja del gráfico: sigue subiendo derechito. El terco nunca deja de
perder clientes, por más turistas que le den. Eso responde A.2c.

```python
import numpy as np
import matplotlib.pyplot as plt

# De cada 100 turistas, cuántos harían clic en cada anuncio.
# OJO: esto el sistema NO lo sabe; lo tiene que ir descubriendo.
tasas = np.array([0.20, 0.35, 0.55, 0.15])
nombres = ["Playa", "Volcanes", "Aventura", "Cultura"]
TURISTAS, REPETICIONES = 4000, 200
mejor = int(np.argmax(tasas))

def campana(estilo, rng):
    clics = np.zeros(4)
    mostrado = np.zeros(4)
    a, b = np.ones(4), np.ones(4)  # "corazonada" sobre cada anuncio
    perdidos = np.zeros(TURISTAS)
    for t in range(TURISTAS):
        if estilo == "terco":
            promedios = np.where(mostrado > 0, clics / np.maximum(mostrado, 1), 1.0)
            elegido = int(rng.integers(4)) if rng.random() < 0.1 else int(np.argmax(promedios))
        else:  # el apostador
            elegido = int(np.argmax(rng.beta(a, b)))
        hizo_clic = rng.random() < tasas[elegido]
        mostrado[elegido] += 1
        clics[elegido] += hizo_clic
        a[elegido] += hizo_clic
        b[elegido] += (not hizo_clic)
        perdidos[t] = tasas[mejor] - tasas[elegido]
    return np.cumsum(perdidos), mostrado

rng = np.random.default_rng(2026)
res = {}
for estilo in ["terco", "apostador"]:
    corridas = [campana(estilo, rng) for _ in range(REPETICIONES)]
    res[estilo] = (
        np.mean([c[0] for c in corridas], axis=0),
        np.mean([c[1] for c in corridas], axis=0)
    )

for estilo, etiqueta in [("terco", "EL TERCO"), ("apostador", "EL APOSTADOR")]:
    perd, most = res[estilo]
    print(f"{etiqueta}")
    print(f"  Clientes perdidos al final: {perd[-1]:.0f}")
    print(f"  Turistas que vieron el mejor anuncio: {100 * most[mejor] / TURISTAS:.1f}%")
    print("  Reparto: " + ", ".join(f"{n} {100*m/TURISTAS:.0f}%" for n, m in zip(nombres, most)))
    print()

dif = res["terco"][0][-1] - res["apostador"][0][-1]
print(f"Diferencia: el terco perdió {dif:.0f} clientes más que el apostador.")

fig, axs = plt.subplots(1, 2, figsize=(11, 3.6))
for estilo, et, c in [("terco", "El terco", "#C0392B"), ("apostador", "El apostador", "#2E8B57")]:
    axs[0].plot(res[estilo][0], color=c, lw=2.5, label=et)
axs[0].set_xlabel("Turistas atendidos")
axs[0].set_ylabel("Clientes perdidos")
axs[0].set_title("Cuánto cuesta decidir mal")
axs[0].legend()
axs[0].grid(alpha=0.3)

x = np.arange(4)
anc = 0.36
for i, (estilo, et, c) in enumerate([("terco", "El terco", "#C0392B"), ("apostador", "El apostador", "#2E8B57")]):
    axs[1].bar(x + (i - 0.5) * anc, 100 * res[estilo][1] / TURISTAS, anc, label=et, color=c)
axs[1].set_xticks(x)
axs[1].set_xticklabels(nombres)
axs[1].set_ylabel("% de turistas")
axs[1].set_title("A quién le mostró cada anuncio")
axs[1].legend()
axs[1].grid(alpha=0.3, axis="y")

plt.tight_layout()
plt.show()
```

### Celda 2: el embalse

### Lean con cuidado: aquí está lo importante de la clase
Con la perilla en 0.30, el sistema decide usar el agua cuando el embalse está a la mitad, y
termina pasando el 60% del tiempo en riesgo de apagones.
Con la perilla en 0.90, en esa misma situación decide quemar búnker: gasta más este mes
a propósito. Y termina pasando el 0% del tiempo en riesgo. Nunca se queda sin agua.
Es el mismo embalse, la misma información y el mismo programa. Lo único que cambió fue
a qué plazo le pedimos que piense. Y fíjense en el gráfico: el cambio no es gradual, hay
un salto alrededor de 0.81. Antes de ese punto el sistema es cortoplacista; después, previsor.


```python
import numpy as np
import matplotlib.pyplot as plt

# Ejemplo DIDACTICO. No es un modelo real del ICE: los numeros son inventados.
EST = ["bajo", "medio", "alto"]
ACC = ["usar agua", "quemar bunker"]

# A donde queda el embalse el mes siguiente segun lo que se decida (incluye la lluvia)
P = np.array([
    [[0.80, 0.20, 0.00], [0.90, 0.10, 0.00], [0.00, 0.50, 0.50]],  # usar agua
    [[0.40, 0.60, 0.00], [0.00, 0.30, 0.70], [0.00, 0.00, 1.00]]   # quemar bunker
])

# Que tan bien le va este mes con cada decision
R = np.array([
    [-100.0, 10.0, 10.0],  # usar agua: barato, salvo si el embalse esta bajo
    [2.0, 2.0, 2.0]        # quemar bunker: siempre alcanza, pero caro
])

def mejor_plan(plazo):
    """plazo = la perilla: 0 es 'solo me importa hoy', casi 1 es 'me importa el futuro'"""
    V = np.zeros(3)
    for _ in range(5000):
        opciones = R + plazo * (P @ V)  # lo de hoy + lo que viene despues
        Vn = opciones.max(axis=0)
        if np.abs(Vn - V).max() < 1e-13:
            V = Vn
            break
        V = Vn
    return opciones.argmax(axis=0)

def a_la_larga(plan):
    """en que situacion termina viviendo el sistema si sigue siempre este plan"""
    M = np.array([P[plan[s], s] for s in range(3)])
    val, vec = np.linalg.eig(M.T)
    p = np.real(vec[:, np.argmin(np.abs(val - 1))])
    return p / p.sum()

for plazo, etiqueta in [(0.30, "PIENSA SOLO EN ESTE MES"), (0.90, "PIENSA EN TODO EL ANIO")]:
    plan = mejor_plan(plazo)
    reparto = a_la_larga(plan)
    print(f"{etiqueta} (perilla del plazo = {plazo})")
    for s in range(3):
        print(f"  si el embalse esta {EST[s]:<6} -> {ACC[plan[s]]}")
    print("  a la larga pasa: " + ", ".join(f"{100*p:.0f}% {e}" for p, e in zip(reparto, EST)))
    print(f"  >>> tiempo en riesgo de apagones: {100*reparto[0]:.0f}%")
    print()

plazos = np.linspace(0.05, 0.99, 200)
riesgo = [100 * a_la_larga(mejor_plan(g))[0] for g in plazos]
fig, ax = plt.subplots(figsize=(6.2, 3.3))
ax.plot(plazos, riesgo, color="#FF8C00", lw=2.6)
ax.fill_between(plazos, 0, riesgo, color="#FFF5E6")
ax.set_xlabel("qué tanto le importa el futuro (la perilla del plazo)")
ax.set_ylabel("% del tiempo en riesgo\nde apagones")
ax.set_title("Pensar más lejos reduce el riesgo")
ax.grid(alpha=0.3)
for g in [0.30, 0.90]:
    ax.axvline(g, color="#6E6E73", ls="--", lw=1.2)
    ax.text(g, 62, f"{g}", ha="center", fontsize=9)
plt.tight_layout()
plt.show()
```
